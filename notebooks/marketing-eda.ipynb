```python
def load_csv_files(path):
    dataset_path = Path(path)
    csv_files = list(dataset_path.rglob("*.csv"))

    dataframes = {}

    for file in csv_files:
        name = file.stem
        dataframes[name] = pd.read_csv(file)

    return dataframes
```

```python
def calculate_kpis(df):
    df = df.copy()

    df['CTR'] = df['Clicks']/df['Impressions']

    df['Lead_Rate'] = df['Leads'] / df['Clicks']

    df['Conversion_Rate'] = df['Conversions'] / df['Leads']

    df['Click_to_Conversion'] = (df['Conversions'] / df['Clicks'])

    df['CPC'] = (df['Acquisition_Cost'] / df['Clicks'])

    df['CPL'] = (df['Acquisition_Cost'] / df['Leads'])

    df['Cost_Per_Conversion'] = (df['Acquisition_Cost'] / df['Conversions'])

    df['Revenue_Per_Conversion'] = (df['Revenue'] / df['Conversions'])

    df['ROAS'] = (df['Revenue'] / df['Acquisition_Cost'])

    return df
```

```python
nykaa_camp_df = calculate_kpis(nykaa_camp_df)
purplle_camp_df = calculate_kpis(purplle_camp_df)
tira_camp_df = calculate_kpis(tira_camp_df)
```

```python
def explode_channels(df):

    channel_df = df.copy()

    channel_df['Channel_Used'] = (channel_df['Channel_Used'].str.split(','))

    channel_df = channel_df.explode('Channel_Used')
    channel_df['Channel_Used'] = (channel_df['Channel_Used'].str.strip())

    return channel_df
```

```python
nykaa_channel = explode_channels(nykaa_camp_df)
purplle_channel = explode_channels(purplle_camp_df)
tira_channel = explode_channels(tira_camp_df)
```

```python
def channel_performance(df):
    
    channel_df = df.groupby('Channel_Used').agg(
        Impressions=('Impressions', 'sum'),
        Clicks=('Clicks', 'sum'),
        Leads=('Leads', 'sum'),
        Conversions=('Conversions', 'sum'),
        Revenue=('Revenue', 'sum'),
        Cost=('Acquisition_Cost', 'sum'),
        ROI=('ROI', 'mean')
    )

    channel_df['CTR'] = (
        channel_df['Clicks'] / channel_df['Impressions']
    )

    channel_df['Conversion_Rate'] = (
        channel_df['Conversions'] / channel_df['Leads']
    )

    channel_df = channel_df[
        [
            'Impressions',
            'Clicks',
            'CTR',
            'Leads',
            'Conversions',
            'Conversion_Rate',
            'Revenue',
            'Cost',
            'ROI'
        ]
    ]

    return channel_df.sort_values(
        'Revenue',
        ascending=False
    )
```

```python
def performance_by_group(df, group_col):
    
    performance = df.groupby(group_col).agg(
        Impressions=('Impressions', 'sum'),
        Clicks=('Clicks', 'sum'),
        Leads=('Leads', 'sum'),
        Conversions=('Conversions', 'sum'),
        Revenue=('Revenue', 'sum'),
        Acquisition_Cost=('Acquisition_Cost', 'sum')
    )

    performance['CTR'] = (
        performance['Clicks'] /
        performance['Impressions']
    )

    performance['Conversion_Rate'] = (
        performance['Conversions'] /
        performance['Leads']
    )

    performance['ROI'] = (
        (performance['Revenue'] - performance['Acquisition_Cost']) /
        performance['Acquisition_Cost']
    )

    performance = performance[
        [
            'Impressions',
            'CTR',
            'Leads',
            'Conversion_Rate',
            'Revenue',
            'ROI',
            'Acquisition_Cost'
        ]
    ]

    return performance.sort_values(
        'Revenue',
        ascending=False
    )
```

```python
def customer_segment_analysis(company_dfs):
    
    results = []

    for company_name, df in company_dfs:
        
        segment = df.groupby('Target_Audience').agg(
            Revenue=('Revenue', 'sum'),
            Conversions=('Conversions', 'sum'),
            Leads=('Leads', 'sum'),
            Acquisition_Cost=('Acquisition_Cost', 'sum')
        )


        segment['Conversion_Rate'] = (
            segment['Conversions'] / segment['Leads']
        )

        segment['ROI'] = (
            (segment['Revenue'] - segment['Acquisition_Cost'])
            / segment['Acquisition_Cost']
        )

        max_revenue = segment['Revenue'].idxmax()
        best_conversion = segment['Conversion_Rate'].idxmax()
        highest_roi = segment['ROI'].idxmax()
        highest_cost = segment['Acquisition_Cost'].idxmax()

        results.append({
            'Company': company_name,
            'Max Revenue Segment': max_revenue,
            'Best Converting Segment': best_conversion,
            'Highest ROI Segment': highest_roi,
            'Highest Acquisition Cost Segment': highest_cost
        })

    return pd.DataFrame(results)
```

```python
companies = [
    ('Nykaa', nykaa_camp_df),
    ('Purplle', purplle_camp_df),
    ('Tira', tira_camp_df)
]
```

```python
customer_segment_analysis(companies)
```
